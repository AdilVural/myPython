```python
# --- 1. Importeer Bibliotheken ---
import pandas as pd
import numpy as np
import os # Nieuwe import voor padmanipulatie
print("Stap 1: Bibliotheken geïmporteerd.")

# --- 2. Laad de Dataset (Aangepast en Robuuster) ---
# Detecteer de huidige werkmap van het script
current_directory = os.getcwd()
print(f"\nDe huidige werkmap is: {current_directory}")
file_name = 'college_statistics.csv'
file_path = os.path.join(current_directory, file_name) # Bouw een veilig pad naar het bestand
print(f"Stap 2: Proberen het bestand te laden vanaf: {file_path}")
# Definieer de kolomnamen handmatig
column_names = ['University', 'Private', 'Apps', 'Accept', 'Enroll', 'Top10perc', 'Top25perc', 'F.Undergrad', 'P.Undergrad', 'Outstate', 'Room.Board', 'Books', 'Personal', 'PhD', 'Terminal', 'S.F.Ratio', 'perc.alumni', 'Expend', 'Grad.Rate']
try:
    # Laad de dataset met de opgegeven kolomnamen
    df_college = pd.read_csv(file_path, header=None, names=column_names)
    print("Stap 2: Dataset succesvol geladen.")
    print(f"Vorm van de DataFrame (rijen, kolommen): {df_college.shape}")
    print("\nDataset informatie:")
    print(df_college.columns)  # Toon de kolomnamen
    print(df_college.describe())  # Toon de beschrijvende statistieken
    
    print("\nEerste 5 rijen van de dataset:")
    print(df_college.head()) # Toon de eerste 5 rijen van de DataFrame

except FileNotFoundError:
    print(f"Fout: Het bestand '{file_name}' is niet gevonden in de map: {current_directory}")
    df_college = pd.DataFrame()  # Lege DataFrame als fallback    
except Exception as e:
    print(f"Fout bij het laden van de dataset: {e}")
    df_college = pd.DataFrame()  # Lege DataFrame als fallback
```

Welke variabele modelleren we?
    In onze analyse modelleren we de variabele Grad.Rate (Afstudeerpercentage). Dit is de afhankelijke variabele. We proberen te begrijpen welke factoren (onafhankelijke variabelen zoals 'PrivateNumeric', 'AcceptRate', 'log_Apps', 'Expend', enz.) een impact hebben op het afstudeerpercentage van een universiteit en hoe we dit percentage kunnen voorspellen.

Korte toelichting over Logistische Regressie (voor het geval je het in de toekomst nodig hebt):
Logistische Regressie is ook een regressiemodel, maar het wordt gebruikt wanneer de afhankelijke variabele een categorische variabele is (bijvoorbeeld binair, zoals "Ja/Nee", "True/False", "0/1"). Het model voorspelt dan de waarschijnlijkheid dat de afhankelijke variabele tot een bepaalde categorie behoort, in plaats van een directe numerieke waarde. Omdat Grad.Rate een percentage (continue variabele) is, is Lineaire Regressie hier de juiste keuze.


```python
# --- Python Code voor Vraag 6a: Doel van Lineaire Regressie en Te Modelleren Variabele ---
# Importeer Matplotlib en Seaborn voor visualisatie (als ze nog niet geïmporteerd zijn)
import matplotlib.pyplot as plt
import seaborn as sns
# Zorg ervoor dat 'df' beschikbaar is vanuit de eerdere stappen
# Als je dit blok apart draait, moet je df_college eerst laden zoals eerder beschreven:
# df_college = pd.read_csv(...)
# df_college = df_college.set_index('University')
# df = df_college.copy() # Maak een kopie om aan te werken
print("\n--- Analyse voor Vraag 6a: Doel van Lineaire Regressie en Te Modelleren Variabele ---")
# 1. Definieer de afhankelijke (target) variabele
target_variable_name = 'Grad.Rate'
print(f"De afhankelijke variabele die we modelleren is: '{target_variable_name}'")
# 2. Toon het type van deze variabele en enkele beschrijvende statistieken
if target_variable_name in df_college.columns:
    print(f"\nDataType van '{target_variable_name}': {df_college[target_variable_name].dtype}")
    print(f"Beschrijvende statistieken van '{target_variable_name}':")
    print(df_college[target_variable_name].describe())
    # Controleer of het een numerieke variabele is
    if pd.api.types.is_numeric_dtype(df_college[target_variable_name]):
        print(f"\nBevestiging: '{target_variable_name}' is een numerieke, continue variabele.")
        print("Dit maakt het geschikt voor Lineaire Regressie.")
        # 3. Visualiseer de distributie van de target variabele
        plt.figure(figsize=(10, 6))
        sns.histplot(df_college[target_variable_name], kde=True, bins=30)
        plt.title(f'Distributie van het Afstudeerpercentage ({target_variable_name})')
        plt.xlabel('Afstudeerpercentage')
        plt.ylabel('Frequentie')
        plt.grid(axis='y', alpha=0.75)
        plt.show()
        print(f"\nHet histogram toont de verdeling van '{target_variable_name}'.")
        print("De spreiding van waarden bevestigt het continue karakter van de variabele.")
    else:
        print(f"\nWaarschuwing: '{target_variable_name}' is geen numerieke variabele. ")
        print("Lineaire Regressie is mogelijk niet de juiste techniek als de target variabele categorisch is.")
else:
    print(f"\nFout: De variabele '{target_variable_name}' is niet gevonden in de DataFrame 'df'.")
    print("Controleer of de kolomnaam correct is en of de data correct is geladen.")
print("\n--- Einde Analyse Vraag 6a ---")
```
